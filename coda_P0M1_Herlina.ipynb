```python
'''
=================================================
Milestone 1

Nama  : Herlina
Batch : CODA-RMT-018

Program ini dibuat untuk mengambil informasi dari website dan diterjemahkan dalam sebuah data 
=================================================
'''

# data manipulation or dataframe
import pandas as pd

# webscraping
from selenium import webdriver
from bs4 import BeautifulSoup
import time
```

```python
# empty list
names = []
brands =[]
prices = []
imgs = []
diskon = []

# create driver
driver = webdriver.Safari()

# loop certain pages
for num_page in range(1, 5):
    # link
    url = f'https://www.kickavenue.com/search/sneakers?sex=F'

    # access the website
    driver.get(url)

    # delay / time sleep
    time.sleep(5)

    # Scroll
    max_scroll = 20
    count = 0

    while len(names) < 50 and count < max_scroll:
        driver.execute_script ("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(10)
        count += 1
    
    # fetch html components
    html = driver.page_source

    # parse
    soup = BeautifulSoup(html, "html.parser")

    # get box element
    boxes = soup.find_all('a', {'target':'_self'})

    # looping box
    for box in boxes:

        # get name
        try:
            name = box.find('h2', {'class':'price-tag'})
            names.append(name.get_text())
        except:
            names.append(None)
        
        # get brand
        try:
            brand = box.find('span', {'class':'price-tag brand'})
            brands.append(brand.get_text())
        except:
            brand.append(None)

        # get price
        try:
            price = box.find('span', {'class':'currency-tag currency-tag__black text-left price-tag text-bold'})
            prices.append(price.get_text())
        except:
            prices.append(None)

        # get image url
        try:
            img = box.find('img', {'class':'card-img'})
            imgs.append(img['src'])
        except:
            imgs.append(None)
        
        # get diskon 
        try:
            el_diskon = box.find('span', {'class':'currency-tag currency-tag__default text-left price-tag slashed'})
            diskon.append(el_diskon.get_text())
        except:
            diskon.append(None)


driver.quit()
```

```python
data = pd.DataFrame(
    {
        'nama produk':names,
        'brand produk':brands,
        'harga produk':prices,
        'link gambar':imgs,
        'diskon':diskon
    }
)

data.info()
```

```python
# menghilangkan IDR dalam Harga produk
df['harga produk'] = df['harga produk'].str.replace('IDR', '', regex=False ) 
# menghilangkan tanda koma (,) pada penulisan harga produk
df['harga produk'] = df['harga produk'].str.replace(',', '', regex=False) 
# mengubah type harga produk yang awalnya str ke float
df['harga produk'] = df['harga produk'].astype(float)
```

```python
df
```

```python
# menghilangkan IDR dalam diskon
df['diskon'] = df['diskon'].str.replace('IDR', '', regex=False ) 
# menghilangkan tand koma (,) dalam diskon
df['diskon'] = df['diskon'].str.replace(',', '', regex=False) 
# merubah type data yang awalnya str menjadi float/numeric
df['diskon'] = pd.to_numeric(df['diskon'], errors='coerce')
```

```python
df
```

```python
# menghitung persen diskon dari selisih dari diskon dikurangi harga produk
df['diskon persen'] = ((df['diskon'] - df['harga produk']) / df['diskon'] * 100)
# hasilnya kita pilih batasnya 2 angka di belakang koma dan type nya kita buat menjadi float
df['diskon persen'] = df['diskon persen'].round(2).astype(float)
```

```python
# Me-rename nama kolom agar lebih mudah di pahami user
df.rename(columns={
    'harga produk': 'harga retail',
    'diskon': 'harga original',
    'diskon persen': 'diskon'
}, inplace = True)
```
